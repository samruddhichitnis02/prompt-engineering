- Each technique contains two separate prompt examples. The outputs are compared based on clarity, correctness, reasoning quality, and usefulness.

### Connecting with the Ollama Model

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3.2",
    temperature=0
)

print("Ollama model connected.")
```

```python
response = llm.invoke("What is 2 + 2?")
print(response.content)
```

## 1. Zero-Shot Prompting

Zero-shot prompting asks the model to complete a task without providing
any examples in the prompt.

```python
zero_shot_prompt_1 = """
Solve this math problem:

A train travels 180 miles in 3 hours.
What is its average speed in miles per hour?
"""

response_1 = llm.invoke(zero_shot_prompt_1)

print(response_1.content)
```

```python
zero_shot_prompt_2 = """
Classify the following sentence as positive, negative, or neutral:

"The delivery arrived earlier than expected and the packaging was excellent."
"""

response_2 = llm.invoke(zero_shot_prompt_2)

print(response_2.content)
```

### Zero-Shot Findings

In the first example, the model correctly calculated the average speed
as 60 miles per hour.

In the second example, the model correctly classified the sentence as
positive because it described an early delivery and excellent packaging.

The model solved both tasks without receiving any examples. However,
the responses included explanations even though the prompts only asked
for an answer or classification.

## 2. Few-Shot Prompting

Few-shot prompting provides examples so that the model can understand
the expected pattern before answering a new question.

```python
few_shot_prompt_1 = """
Solve the examples by calculating distance divided by time.

Example 1:
Distance = 100 miles
Time = 2 hours
Answer = 50 miles per hour

Example 2:
Distance = 150 miles
Time = 3 hours
Answer = 50 miles per hour

Now solve this problem:
Distance = 240 miles
Time = 4 hours
Answer:
"""

response_1 = llm.invoke(few_shot_prompt_1)

print(response_1.content)
```

```python
few_shot_prompt_2 = """
Classify each sentence as positive or negative.

Example 1:
Sentence: "The movie was enjoyable and inspiring."
Classification: Positive

Example 2:
Sentence: "The service was slow and disappointing."
Classification: Negative

Now classify this sentence:
"The product was easy to use and worked perfectly."
Classification:
"""

response_2 = llm.invoke(few_shot_prompt_2)

print(response_2.content)
```

### Few-Shot Findings

In the first example, the model followed the demonstrated formula and
calculated the answer as 60 miles per hour.

In the second example, the model followed the demonstrated classification
format and classified the sentence as positive.

Compared with zero-shot prompting, few-shot prompting gave the model a
clearer expected format and reduced unnecessary explanation.

## 3. Chain-of-Thought Prompting

Chain-of-thought prompting asks the model to explain the intermediate
steps used to reach the final answer.

```python
cot_prompt_1 = """
Solve the following problem step by step and then provide the final answer.

A store gives a 20% discount on a jacket priced at 80 dollars.
What is the final price after the discount?
"""

response_1 = llm.invoke(cot_prompt_1)

print(response_1.content)
```

```python
cot_prompt_2 = """
Analyze the following logical reasoning problem step by step.

All roses are flowers.
Some flowers fade quickly.

Can we conclude that some roses fade quickly?
Explain your reasoning and provide the final answer.
"""

response_2 = llm.invoke(cot_prompt_2)

print(response_2.content)
```

### Chain-of-Thought Findings

In the first example, the model calculated the discount in separate steps
and correctly found the final price.

In the second example, the model examined each statement and correctly
explained why the conclusion cannot be guaranteed.

Compared with zero-shot and few-shot prompting, chain-of-thought prompting
produced a more detailed explanation of the reasoning process. However, the
response was longer than necessary for a simple problem.

## 4. Zero-Shot Chain-of-Thought Prompting

Zero-shot chain-of-thought prompting does not provide examples.
Instead, it asks the model to reason through the problem step by step.

```python
zero_shot_cot_prompt_1 = """
Think through this problem step by step and provide the final answer:

A classroom has 24 students. One third of the students are absent.
How many students are present?
"""

response_1 = llm.invoke(zero_shot_cot_prompt_1)

print(response_1.content)
```

```python
zero_shot_cot_prompt_2 = """
Let's reason step by step.

A farmer has chickens and cows.
There are 10 animals and 28 legs in total.

Let x be the number of chickens and y be the number of cows.

Use these two equations:
x + y = 10
2x + 4y = 28

Solve the equations and give the number of chickens and cows.
"""

response_2 = llm.invoke(zero_shot_cot_prompt_2)

print(response_2.content)
```

### Zero-Shot CoT Findings

In the first example, the model correctly calculated that 16 students
were present.

In the second example, the model used the two equations and correctly
calculated that there were 6 chickens and 4 cows.

Compared with regular zero-shot prompting, zero-shot CoT produced a
step-by-step explanation without needing examples. The reasoning was
more detailed, but the response was also longer.

## 5. Meta-Prompting

Meta-prompting asks the model to create, improve, or evaluate a prompt
instead of directly solving the original task.

```python
meta_prompt_1 = """
You are a prompt engineering expert.

Create a clear prompt that asks a language model to summarize
a scientific article for a beginner. The prompt should request
simple language, the main idea, important findings, and a short
conclusion.
"""

response_1 = llm.invoke(meta_prompt_1)

print(response_1.content)
```

```python
meta_prompt_2 = """
You are a prompt engineering expert.

Improve this weak prompt so that the model gives a concise and
well-organized answer:

"Tell me about climate change."

Return only the improved prompt.
"""

response_2 = llm.invoke(meta_prompt_2)

print(response_2.content)
```

### Meta-Prompting Findings

In the first example, the model created a structured prompt with clear
instructions about the audience, content, tone, and length.

In the second example, the model improved a vague prompt by specifying
the topics that should be covered, such as causes, effects, solutions,
and mitigation strategies.

Meta-prompting is useful because the model helps design better prompts
instead of directly answering the original task.

## 6. Tree of Thoughts Prompting

Tree of Thoughts prompting asks the model to consider multiple possible
solutions, compare them, and select the most appropriate one.

```python
tree_prompt_1 = """
Solve this problem using a tree of thoughts approach.

You have 20 dollars.
Each notebook costs 4 dollars.
Each pen costs 2 dollars.
You must buy at least 2 notebooks.

Consider every possible number of notebooks from 2 through 5.
For each case, calculate:
1. The notebook cost.
2. The remaining money.
3. The maximum number of pens.
4. The total number of items.

Compare all valid cases and select the combination with the greatest
total number of items. Show the calculations and give the final answer.
"""

response_1 = llm.invoke(tree_prompt_1)

print(response_1.content)
```

```python
tree_prompt_2 = """
Use a tree of thoughts approach to choose the best way to study for
an examination.

Compare these three options:
A. Study only from lecture notes.
B. Review lecture notes and solve practice questions.
C. Watch videos without reviewing notes.

For each option:
1. Explain one advantage.
2. Explain one disadvantage.
3. Evaluate its effectiveness.

Then compare all three options and select the best study strategy.
"""

response_2 = llm.invoke(tree_prompt_2)

print(response_2.content)
```

### Tree of Thoughts Findings

In the first example, the model explored different combinations of
notebooks and pens. It calculated the cost and total number of items for
each possible case, then selected 2 notebooks and 6 pens as the best
combination, giving a total of 8 items.

In the second example, the model evaluated three study strategies by
comparing their advantages, disadvantages, and effectiveness. It selected
reviewing lecture notes and solving practice questions as the best option.

Tree of Thoughts allowed the model to compare multiple possible paths
before selecting a final answer. It provided a more complete analysis than
the other techniques, but the response was longer and required clearer
instructions to consider every possible case.

## Overall Comparison

| Technique | How the prompt was written | Output behavior |
|---|---|---|
| Zero-Shot | Only the task was provided | The model solved the task but sometimes gave extra explanation |
| Few-Shot | Examples were provided before the new task | The model followed the demonstrated pattern and format |
| Chain-of-Thought | The model was asked to explain the solution step by step | The responses were detailed and showed intermediate reasoning |
| Zero-Shot CoT | No examples were provided, but step-by-step reasoning was requested | The model reasoned through the problem without examples |
| Meta-Prompting | The model was asked to create or improve another prompt | The model produced clearer and more structured prompts |
| Tree of Thoughts | The model was asked to explore and compare multiple solutions | The model considered several possibilities and selected the best one |

### Overall Findings

Zero-shot prompting was the simplest approach but provided the least
guidance. Few-shot prompting helped the model follow a specific pattern.
Chain-of-thought and zero-shot chain-of-thought produced more detailed
reasoning, although their responses were longer.

Meta-prompting was useful for creating and improving prompts. Tree of
Thoughts provided the most extensive comparison by examining multiple
possible solutions, but it also produced the longest output and initially
missed one possible combination until the prompt was made more specific.
